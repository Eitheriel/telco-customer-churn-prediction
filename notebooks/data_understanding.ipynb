During the initial inspection, we identified that the Total Charges feature is stored as an object instead of a numeric data type. Further investigation revealed that this is caused by 11 empty string values. These records correspond to customers with zero tenure, suggesting that the missing total charges may represent expected business behavior rather than erroneous data. The column will be handled during preprocessing.

## Missing Values

```python
df.isna().sum()
```

Missing values were detected only in the Churn Reason feature. Since this information is available only for customers who have churned, the missing values are expected rather than indicating poor data quality.

The Churn Reason feature records the reason why a customer left the company. Since this information becomes available only after the churn event has occurred, it would not be available at prediction time. Using this feature during model training would introduce data leakage, resulting in unrealistically optimistic model performance.

## Cardinality

```python
df.nunique()
```

Features with a cardinality of one do not contain any variability and therefore provide no discriminative information for the model. These features will be evaluated for removal during preprocessing.

## Numeric Statistics

The numerical features appear to contain valid values within reasonable ranges. ## Categorical Statistics

```python
df.describe(include="string")
```

The categorical features contain plausible values and no unexpected categories were identified during the initial inspection. Several features, such as Country and State, contain only a single unique value and are therefore unlikely to provide predictive information.

# Key Findings

The initial inspection of the dataset revealed several important observations:

- The dataset contains 7,043 customer records and 33 features.
- The `Total Charges` feature is stored as an object due to 11 empty string values associated with customers having zero tenure.
- Missing values occur only in the `Churn Reason` feature and appear to be expected.
- Several features (`Count`, `Country`, and `State`) have no variability and will be considered for removal during preprocessing.
- The remaining numerical and categorical features appear consistent based on the initial inspection.

## Feature Assessment

| Feature | Role | Recommendation |
|---------|------|--------------------|
| CustomerID | Identifier | Remove before training |
| Count | Constant feature | Candidate for removal |
| Country | Constant feature | Candidate for removal |
| State | Constant feature | Candidate for removal |
| Total Charges | Numerical feature | Convert to numeric during preprocessing |
| Churn Label | Target variable | Prediction target |
| Churn Value | Encoded target | Redundant with `Churn Label` |
| Churn Score | Unknown | Investigate before modeling |
| CLTV | Business metric | Investigate before modeling |
| Churn Reason | Post-event information | Exclude to prevent data leakage |
