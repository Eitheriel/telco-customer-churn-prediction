```python
import pandas as pd
import numpy as np

df = pd.read_excel("../data/raw/telco_customer_churn.xlsx")
```

```python
# Create a copy of the original DataFrame to work with
df_clean = df.copy()
```

# Convert Total Charges to numeric

```python
df_clean["Total Charges"] = pd.to_numeric(
    df_clean["Total Charges"],
    errors="coerce"
)

df_clean.loc[
    (df_clean["Total Charges"].isna()) &
    (df_clean["Tenure Months"] == 0),
    "Total Charges"
] = 0

df_clean["Total Charges"].isna().sum()
```

```python
df_clean["Total Charges"].dtype
```

# Remove unnecessary columns

## Remove constant features

```python
# Remove unnecessary columns
constant_columns = ["Count", "Country", "State"]

df_clean = df_clean.drop(columns=constant_columns)
```

## Remove identifier and leakage features

```python
# Remove columns that are not needed for analysis
columns_to_remove = [
    "CustomerID",
    "Churn Value",
    "Churn Score",
    "Churn Reason"
]

df_clean = df_clean.drop(columns=columns_to_remove)

# Remove the "Lat Long" column as it is redundant
df_clean = df_clean.drop(columns=["Lat Long"])
```

The preprocessing steps successfully converted the Total Charges feature to a numerical data type and removed constant, redundant, identifier, and post-event features identified during the data understanding phase. The resulting dataset contains 25 features with no missing values and is ready for exploratory data analysis.

```python
# Save preprocessed dataset
df_clean.to_csv(
    "../data/processed/telco_customer_churn_cleaned.csv",
    index=False
)
```
